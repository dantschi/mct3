# Aktivitätserkennung aus echten Messdaten
Im MCT3-Repository ausführen. Benötigt Python 3, NumPy und Matplotlib.
Die MotionSense-Daten stehen unter ihrer mitgelieferten MIT-Lizenz.
Das Skript trainiert 96 → 16 → 3 mit fester Personentrennung und erzeugt Modell, Auswertung und Abbildungen.
Die Normalisierung lernt nur auf Trainingsdaten. Testpersonen bleiben bei der Anpassung ausgeschlossen.


```python
from pathlib import Path
import json, subprocess, sys
repo = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "code/vl08_activity_demo.py").is_file())
subprocess.run([sys.executable, str(repo / "code/vl08_activity_demo.py")],
               cwd=repo, check=True)
summary = json.loads((repo / "assets/data/08_motion_sense/model-summary.json").read_text())
for role, result in summary["metrics_exported_float32_model"].items():
    print(role, f"{result['accuracy']:.2%}")
print("Beispiel:", summary["example"])
```

## Auswerten
Vergleichen Sie Trainings- und Testgüte. Ein großes Gefälle zeigt, dass wiedererkannte Trainingsfenster noch keine allgemeine Aktivitätserkennung belegen.
Die Messdaten haben 50 Hz, Einheit g und enthalten Gravitation. Eigene phyphox-Daten benötigen zuerst denselben Eingabevertrag; dies ist keine bereits ausgeführte Erweiterung.
